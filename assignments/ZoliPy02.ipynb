```python
# Method 2:

# توزیع مقادیر ستون country
distribution = df['country'].value_counts(normalize=True)

# پر کردن NaNها به صورت تصادفی بر اساس توزیع
df.loc[df['country'].isna(), 'country'] = np.random.choice(
    distribution.index,
    size=df['country'].isna().sum(),
    p=distribution.values
)

print(df['country'].value_counts())
```

و تمرین 3:


دیتافریم پاندا 2 تا متد دیگه به نام ffill و bfill هم داره برای پرکردن
NaNها. این دو تا رو بررسی کنید و مثال بزنید و اثرش رو ببینید.


متدهای ffill و bfill برای پر کردن مقادیر NaN استفاده می‌شوند.


متد ffill مقدار NaN را با مقدار قبلی پر می‌کند.

متد bfill مقدار NaN را با مقدار بعدی پر می‌کند.


در مثال زیر اثر هر دو متد را روی یک ستون مشاهده می‌کنیم.



```python

df = pd.DataFrame({
    'age': [20, 25, None, 30, None, 40]
})

print("before")
print(df)

df['age_ffill'] = df['age'].ffill()
df['age_bfill'] = df['age'].bfill()

print("after:")
print(df)
```
