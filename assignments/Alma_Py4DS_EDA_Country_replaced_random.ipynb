```python
import pandas as pd
```

```python
df_org = pd.read_csv("/content/bank_data_with_NaN.csv")
```

```python
df = df_org.copy()

switzerland_count = df["country"].value_counts()["Switzerland"]
germany_count = df["country"].value_counts()["Germany"]
austria_count = df["country"].value_counts()["Austria"]
known_country = df["country"].notna().sum()
unknown_country = df["country"].isna().sum()


print ("Germany_before:", germany_count)
print ("Switzerland_before:", switzerland_count)
print ("Austria_before:", austria_count)
print ("Known_country_before:", known_country)
print ("Unknown_country_before:", unknown_country)


germany_percent = germany_count / known_country
switzerland_percent = switzerland_count / known_country
austria_percent = austria_count / known_country

germany_replace = round(germany_percent * unknown_country)
switzerland_replace = round(switzerland_percent * unknown_country)
austria_replace = round(austria_percent * unknown_country)


nan_cells = df[df["country"].isna()].index
selected_germany = nan_cells.to_series().sample(germany_replace)
df.loc[selected_germany, "country"] = "Germany"

nan_cells = df[df["country"].isna()].index
selected_switzerland = nan_cells.to_series().sample(switzerland_replace)
df.loc[selected_switzerland, "country"] = "Switzerland"

nan_cells = df[df["country"].isna()].index
selected_austria = nan_cells.to_series().sample(austria_replace)
df.loc[selected_austria, "country"] = "Austria"


switzerland_count = df["country"].value_counts()["Switzerland"]
germany_count = df["country"].value_counts()["Germany"]
austria_count = df["country"].value_counts()["Austria"]
known_country = df["country"].notna().sum()


print ("Germany_after:", germany_count)
print ("Switzerland_after:", switzerland_count)
print ("Austria_after:", austria_count)
print ("Known_country_after:", known_country)
```
