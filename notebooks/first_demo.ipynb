# Let's import required packages

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Let's read the dataset

```python
df = pd.read_csv("../datasets/Customer-Churn-Records.csv")

print(df.shape)

df.head()
```

```python
df["Exited"].value_counts()
```

```python
df["Exited"].value_counts(normalize=True)
```

# Some Visualizations

## 1. Bar Chart using `seaborn`

```python
sns.countplot(data=df, x="Exited", color="green")
```

## 2. Pie Chart using `matplotlib.pyplot`

```python
cnt = df["Geography"].value_counts()

plt.figure(figsize=(4, 4))

plt.pie(cnt.values, labels=cnt.index, 
        autopct="%1.2f%%", startangle=45, 
        wedgeprops={"edgecolor": "white", "linewidth": 2}
        )

plt.title("Distribution of Geography")
plt.axis("equal")
plt.show()
```
