```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import StandardScaler
```

```python
df_org = pd.read_csv("/content/breast-cancer.csv")

print(df_org.shape)

df_org.head()
```

```python
df = df_org.copy()

df.isna().sum().sum()
```

```python
# Identify non-numeric columns in the DataFrame
non_numeric_cols = df.select_dtypes(exclude=['number']).columns

print("Non-numeric columns:")
print(list(non_numeric_cols))
```

```python
# Convert 'B' to 0 and 'M' to 1 in the 'diagnosis' column
df['diagnosis'] = df['diagnosis'].map({'B': 0, 'M': 1})

df.head()
```

```python
df["diagnosis"].value_counts(normalize=True)
```

```python
df.drop("id", axis=1, inplace=True)

df.head()
```

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=1234)

print("Train Set Shape:", train_df.shape)
print("Test Set Shape:", test_df.shape)

print("\nTrain DataFrame:")
display(train_df)
```

```python
# جدا کردن ویژگی‌ها و هدف
X = train_df.drop(columns=['diagnosis'])  # All columns except 'Exited'
y = train_df['diagnosis']                 # The target column

X_test = test_df.drop(columns=['diagnosis'])
y_test = test_df['diagnosis']

# ۱. مقیاس‌بندی داده‌ها (StandardScaler)
scaler = StandardScaler()

# فیت کردن فقط روی داده‌های آموزشی و تبدیل آن‌ها
X_train_scaled = scaler.fit_transform(X)

# تبدیل داده‌های تست با همان اسکیلر آموزش‌دیده
X_test_scaled = scaler.transform(X_test)

# ۲. ساخت و آموزش مدل KNN با داده‌های مقیاس‌شده
clf_knn = KNeighborsClassifier(n_neighbors=7)
clf_knn.fit(X_train_scaled, y)

# ۳. پیش‌بینی روی داده‌های تست مقیاس‌شده
y_pred = clf_knn.predict(X_test_scaled)

# ۴. ارزیابی مدل
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
clf_lr = LogisticRegression(random_state=1234, class_weight={0:1, 1:10})
clf_lr.fit(X_train_scaled, y)

y_pred = clf_lr.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
